This notebook covers:

- Inspecting shape, columns, dtypes
- Data quality report (missing values, duplicates)
- Sample row inspection

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style='whitegrid')

from src.data_loader import load_raw_orders, generate_data_quality_report, print_data_quality_report
```

```python
# Load raw data
df_raw = load_raw_orders()
print(f'Shape: {df_raw.shape}')
df_raw.head(5)
```

```python
# Column names and types
print('Columns and dtypes:')
for col, dtype in df_raw.dtypes.items():
    print(f'  {col:<35} {dtype}')
```

```python
# Data quality report
report = generate_data_quality_report(df_raw)
print_data_quality_report(report)
```

```python
# Date range
print('Order Date range:', df_raw['Order Date'].min(), 'to', df_raw['Order Date'].max())
print('Ship Date range: ', df_raw['Ship Date'].min(), 'to', df_raw['Ship Date'].max())
```

```python
# Unique values in categorical columns
cat_cols = ['Category', 'Sub-Category', 'Segment', 'Region', 'Ship Mode']
for col in cat_cols:
    if col in df_raw.columns:
        print(f'\n{col}: {sorted(df_raw[col].unique())}')
```

```python
# Numerical summary
df_raw[['Sales','Quantity','Discount','Profit']].describe().round(2)
```

```python
# Sales and Profit distributions
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(df_raw['Sales'], bins=60, color='steelblue', edgecolor='white')
axes[0].set_title('Sales Distribution (raw)')
axes[0].set_xlabel('Sales (USD)')
axes[1].hist(df_raw['Profit'], bins=60, color='seagreen', edgecolor='white')
axes[1].set_title('Profit Distribution (raw)')
axes[1].set_xlabel('Profit (USD)')
plt.tight_layout()
plt.show()
```

```python
# Discount distribution
plt.figure(figsize=(8,4))
plt.hist(df_raw['Discount'], bins=20, color='darkorange', edgecolor='white')
plt.title('Discount Distribution')
plt.xlabel('Discount (fraction)')
plt.tight_layout()
plt.show()
```
