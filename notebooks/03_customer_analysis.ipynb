```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style='whitegrid', font_scale=1.1)

from src.feature_engineering import load_processed
from src.customer_analysis import (
    customer_revenue_table, compute_rfm, run_customer_analysis
)
```

```python
df = load_processed()
print(f'Loaded {len(df):,} rows')
```

```python
# Customer revenue table (top 20)
cust = customer_revenue_table(df)
print(f'Total customers: {len(cust)}')
cust.head(20)
```

```python
# Revenue distribution
plt.figure(figsize=(8,4))
plt.hist(cust['Total_Sales'], bins=40, color='steelblue', edgecolor='white')
plt.title('Customer Revenue Distribution')
plt.xlabel('Total Revenue (USD)')
plt.tight_layout()
plt.show()
```

```python
# RFM computation
rfm = compute_rfm(df)
print(f'RFM table: {len(rfm)} customers')
rfm.head(10)
```

```python
# RFM segment counts
print(rfm['RFM_Segment'].value_counts())
```

```python
# Revenue by RFM segment
seg_rev = rfm.groupby('RFM_Segment')['Monetary'].sum().sort_values(ascending=False)
fig, ax = plt.subplots(figsize=(8,4))
seg_rev.plot(kind='bar', color='steelblue', ax=ax)
ax.set_title('Total Revenue by RFM Segment')
ax.set_ylabel('Total Sales (USD)')
ax.tick_params(axis='x', rotation=30)
plt.tight_layout()
plt.show()
```

```python
# Scatter: Frequency vs Monetary, coloured by Recency
fig, ax = plt.subplots(figsize=(9,5))
sc = ax.scatter(rfm['Frequency'], rfm['Monetary'], c=rfm['Recency'],
                cmap='RdYlGn_r', alpha=0.65, s=40)
plt.colorbar(sc, ax=ax, label='Recency (days)')
ax.set_xlabel('Frequency (# orders)')
ax.set_ylabel('Monetary Value (USD)')
ax.set_title('RFM Scatter')
plt.tight_layout()
plt.show()
```
